```python
import pandas as pd

df = pd.read_csv('flights_sample_3m.csv')

df.head()
```

```python
min_date = df['FL_DATE'].min()
max_date = df['FL_DATE'].max()
print(f"The dataset covers flights from {min_date} to {max_date}")

df = df.sort_values(by='FL_DATE')

df = df.reset_index(drop=True)

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df['FL_DATE'] = pd.to_datetime(df['FL_DATE'])

print(df['FL_DATE'].dtype)
```

```python
missing_data = df.isnull().sum()

missing_data[missing_data > 0].sort_values(ascending=False)
```

```python
delay_columns = ['DELAY_DUE_CARRIER', 'DELAY_DUE_SECURITY', 'DELAY_DUE_NAS', 'DELAY_DUE_WEATHER', 'DELAY_DUE_LATE_AIRCRAFT']
df[delay_columns] = df[delay_columns].fillna(0)

df['CANCELLATION_CODE'] = df['CANCELLATION_CODE'].fillna('Not Cancelled')

df = df.dropna()

print("Max missing values in any column now:", df.isnull().sum().max())
```

```python
df.describe().round(2)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

sns.histplot(df['DEP_DELAY'], bins=100, kde=False, color='royalblue', ax=axes[0])
axes[0].set_title('Distribution of Departure Delays')
axes[0].set_xlabel('Departure Delay (Minutes)')
axes[0].set_ylabel('Frequency')

sns.boxplot(x=df['DEP_DELAY'], color='lightblue', ax=axes[1])
axes[1].set_title('Boxplot of Departure Delays')
axes[1].set_xlabel('Departure Delay (Minutes)')

plt.tight_layout()
plt.show()
```

```python
filtered_delays = df['DEP_DELAY'][(df['DEP_DELAY'] >= -10) & (df['DEP_DELAY'] <= 120)]

plt.figure(figsize=(10, 6))
sns.histplot(filtered_delays, bins=50, color='teal', kde=True)

plt.title('Zoomed-In Distribution of Departure Delays (0 to 120 Minutes)', fontsize=14)
plt.xlabel('Departure Delay (Minutes)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
airline_delay_stats = df.groupby('AIRLINE')['DEP_DELAY'].mean().reset_index()
airline_delay_stats = airline_delay_stats.sort_values(by='DEP_DELAY', ascending=False)

plt.figure(figsize=(10, 8))
sns.barplot(x='DEP_DELAY', y='AIRLINE', data=airline_delay_stats, palette='coolwarm')

plt.title('Average Departure Delay by Airline', fontsize=14)
plt.xlabel('Average Departure Delay (Minutes)', fontsize=12)
plt.ylabel('Airline', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
df['DAY_OF_WEEK'] = df['FL_DATE'].dt.day_name()

days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

daily_delays = df.groupby('DAY_OF_WEEK')['DEP_DELAY'].mean().reindex(days_order).reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='DAY_OF_WEEK', y='DEP_DELAY', data=daily_delays, palette='viridis', hue='DAY_OF_WEEK', legend=False)

plt.title('Average Departure Delay by Day of the Week', fontsize=14)
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Average Departure Delay (Minutes)', fontsize=12)
plt.xticks(rotation=45) 
plt.tight_layout()
plt.show()
```

### 8. תובנות וסיפור הנתונים

* **הטיה או סיכון בנתונים:**

* **נקודות כשל לשימושים הנדסיים וסטטיסטיים:**

* **תובנות אישיות והסקת מסקנות:**
